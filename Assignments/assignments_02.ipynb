##### Chapter 01:  Git init, git staging, restore stage, commit
```bash
MINGW64 ~/Desktop/DevOps
$ mkdir nanur-khata && cd nanur-khata/

MINGW64 ~/Desktop/DevOps/nanur-khata
$ git init
Initialized empty Git repository in C:/Users/Rafid/Desktop/DevOps/nanur-khata/.git/

MINGW64 ~/Desktop/DevOps/nanur-khata (main)
$ touch bhapa-pitha.md

MINGW64 ~/Desktop/DevOps/nanur-khata (main)
$ git add .

MINGW64 ~/Desktop/DevOps/nanur-khata (main)
$ cat \<< 'end' > bhapa-pitha.md
> Winter is coming. Write the the bhapa pitha recipe first.
> Take one and a half cups of gur first.
> end 

MINGW64 ~/Desktop/DevOps/nanur-khata (main)
$ git diff
index e69de29..0f912ac 100644
--- a/bhapa-pitha.md
+++ b/bhapa-pitha.md
@@ -0,0 +1,2 @@
+Winter is coming. Write the the bhapa pitha recipe first.
+Take one and a half cups of gur first.

MINGW64 ~/Desktop/DevOps/nanur-khata (main)
$ git commit -m "work in progress: gur added"

MINGW64 ~/Desktop/DevOps/nanur-khata (main)
$ touch bazar-list.txt

MINGW64 ~/Desktop/DevOps/nanur-khata (main)
$ git add .

MINGW64 ~/Desktop/DevOps/nanur-khata (main)
$ git status
On branch main
Changes to be committed:
  (use "git restore --staged <file>..." to unstage)
        new file:   bazar-list.txt
        modified:   bhapa-pitha.md

MINGW64 ~/Desktop/DevOps/nanur-khata (main)
$ git restore --staged bazar-list.txt 

MINGW64 ~/Desktop/DevOps/nanur-khata (main)
$ git status
On branch main
Changes to be committed:
  (use "git restore --staged <file>..." to unstage)
        modified:   bhapa-pitha.md

Untracked files:
  (use "git add <file>..." to include in what will be committed)
        bazar-list.txt
```
