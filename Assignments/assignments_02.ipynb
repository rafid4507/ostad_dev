---
###### git fetch origin contacts the remote server and downloads all new commits, branches, and tags into the local copy of the remote branches (e.g., origin/main), but does not touch or modify the active working directory. This allows us to inspect changes (using git log HEAD..origin/main or git diff) before applying them. It is a safer option. we can use git merge origin/main afterwards. But git pull fetch and merge in one step.

###### Task D: reset --hard
![reset_hard](./Assets/delete_branch.png)

---

###### git reset --hard moves the branch pointer backward and deletes commits entirely. If pushed to a shared remote (requiring a --force), any commits teammates based their work on vanish from the branch history, leading to broken tracking branches and missing code.
---
###### git rebase creates entirely brand-new commit hashes for existing commits by replaying them on top of a new base. When pushed to a shared remote branch, everyone else's local copy will diverge from the remote, resulting in duplicate commits, messy merge conflicts, and corrupted history when teammates attempt to git pull.
